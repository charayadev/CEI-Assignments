# Procurement & Vendor Spend Analytics Pipeline

## SQL Business Analytics using Gold Layer

### Project Overview

This notebook performs business analysis using the Gold Layer of the Procurement & Vendor Spend Analytics Pipeline.

The Bronze, Silver, and Gold layers have already been created. Therefore, this notebook focuses only on analyzing the final curated Gold tables to generate meaningful business insights.


```sql
SELECT *
FROM gold_vendor_spend_summary
LIMIT 10;
```

```sql
-- Gold Table 3: Vendor Risk Classification

-- What are we doing?

-- In this step, we are displaying the first 10 records from the `gold_vendor_risk_classification` table.

-- Why are we doing it?

-- Before performing business analysis, it is important to understand the structure and contents of the table. This helps us identify the available columns and understand how vendor risk information is stored.

SELECT *
FROM gold_vendor_risk_classification
LIMIT 10;
```

```sql
-- Gold Table 4: Regional Spend Analysis

-- What are we doing?

-- In this step, we are displaying the first 10 records from the `gold_regional_spend_analysis` table.

-- Why are we doing it?

-- Understanding the structure of the table before performing analysis helps us identify the available columns and the type of regional spending information stored in the Gold layer.


SELECT *
FROM gold_regional_spend_analysis
LIMIT 10;
```

```sql
-- Query 1: Display Total Spend by Vendor

-- What are we doing?

-- In this query, we calculate the total procurement spend for each vendor.


SELECT
    vendor_id,
    SUM(total_spend) AS total_vendor_spend
FROM gold_vendor_spend_summary
GROUP BY vendor_id
ORDER BY total_vendor_spend DESC;
```

```sql
-- Query 2: Top 10 Vendors by Spend

-- What are we doing?

-- This query retrieves the top 10 vendors based on their total procurement spend.

SELECT
    vendor_id,
    SUM(total_spend) AS total_vendor_spend
FROM gold_vendor_spend_summary
GROUP BY vendor_id
ORDER BY total_vendor_spend DESC
LIMIT 10;
```

```sql
-- Query 3: Average Invoice Amount by Vendor

-- What are we doing?

-- This query displays the average invoice amount for each vendor.


SELECT
    vendor_id,
    AVG(average_invoice_amount) AS average_invoice_amount
FROM gold_vendor_spend_summary
GROUP BY vendor_id
ORDER BY average_invoice_amount DESC;
```

```sql
-- Query 4: Find Vendors Whose Spend is Above Average

-- What are we doing?

-- This query identifies vendors whose total procurement spend is greater than the average spend across all vendors.


SELECT
    vendor_id,
    total_spend
FROM gold_vendor_spend_summary
WHERE total_spend > (
    SELECT AVG(total_spend)
    FROM gold_vendor_spend_summary
)
ORDER BY total_spend DESC;
```

```sql
-- Query 5: Display Region-wise Total Spend

-- What are we doing?

-- This query calculates the total procurement spending for each region.


SELECT
    region,
    SUM(total_spend) AS regional_total_spend
FROM gold_regional_spend_analysis
GROUP BY region
ORDER BY regional_total_spend DESC;
```

```sql
-- Query 6: Count Vendors in Each Risk Category

-- What are we doing?

-- This query counts the number of vendors in each risk category.+


SELECT
    vendor_risk,
    COUNT(vendor_id) AS total_vendors
FROM gold_vendor_risk_classification
GROUP BY vendor_risk
ORDER BY total_vendors DESC;
```

```sql
SELECT
    vendor_id,
    average_variance_percentage,
    has_overdue_invoice,
    vendor_risk
FROM gold_vendor_risk_classification
WHERE vendor_risk = 'High Risk';
```

```sql
-- Query 8: Display Overcharged Purchase Orders

-- What are we doing?

-- This query identifies all purchase orders where the actual amount paid is higher than the expected amount.
SELECT
    po_id,
    vendor_id,
    item_name,
    expected_amount,
    actual_amount,
    variance,
    variance_percentage
FROM gold_price_variance_report
WHERE price_status = 'Overcharged'
ORDER BY variance DESC;
```

```sql
-- Query 9: Display Undercharged Purchase Orders

--  What are we doing?

-- This query retrieves all purchase orders where the actual amount paid is lower than the expected amount.


SELECT
    po_id,
    vendor_id,
    item_name,
    expected_amount,
    actual_amount,
    variance,
    variance_percentage
FROM gold_price_variance_report
WHERE price_status = 'Undercharged'
ORDER BY variance;
```

```sql
-- Query 10: Count Purchase Orders by Region

-- What are we doing?

-- This query counts the total number of purchase orders for each region.


SELECT
    region,
    SUM(total_purchase_orders) AS total_purchase_orders
FROM gold_regional_spend_analysis
GROUP BY region
ORDER BY total_purchase_orders DESC;
```

```sql
-- Query 11: Find Maximum Vendor Spend

-- What are we doing?

-- This query finds the highest procurement spend among all vendors.

SELECT
    MAX(total_spend) AS maximum_vendor_spend
FROM gold_vendor_spend_summary;
```

```sql
-- Query 12: Find Minimum Vendor Spend

-- What are we doing?

-- This query finds the minimum procurement spend among all vendors.

SELECT
    MIN(total_spend) AS minimum_vendor_spend
FROM gold_vendor_spend_summary;
```

```sql
-- Query 13: Display Vendor Spend Category using CASE WHEN

-- What are we doing?

-- This query classifies vendors into different spending categories using the `CASE WHEN` statement.
SELECT
    vendor_id,
    total_spend,
    CASE
        WHEN total_spend > 500000 THEN 'High Spend'
        WHEN total_spend BETWEEN 200000 AND 500000 THEN 'Medium Spend'
        ELSE 'Low Spend'
    END AS spend_category
FROM gold_vendor_spend_summary
ORDER BY total_spend DESC;
```

```sql
-- Query 14: Vendors Spending Above Average using a CTE

-- What are we doing?

-- This query uses a Common Table Expression (CTE) to calculate the average vendor spend and then displays vendors whose spending is above that average.


WITH average_spend AS (
    SELECT AVG(total_spend) AS avg_spend
    FROM gold_vendor_spend_summary
)

SELECT
    vendor_id,
    total_spend
FROM gold_vendor_spend_summary, average_spend
WHERE total_spend > avg_spend
ORDER BY total_spend DESC;
```

```sql
-- Query 15: Final Business Summary

-- What are we doing?

-- This query generates an overall procurement summary using the Gold Layer tables.


    SELECT
    (SELECT COUNT(DISTINCT vendor_id)
     FROM gold_vendor_spend_summary) AS total_vendors,

    (SELECT SUM(total_spend)
     FROM gold_vendor_spend_summary) AS total_spend,

    (SELECT AVG(total_spend)
     FROM gold_vendor_spend_summary) AS average_spend,

    (SELECT COUNT(*)
     FROM gold_vendor_risk_classification
     WHERE vendor_risk = 'High Risk') AS high_risk_vendors,

    (SELECT COUNT(*)
     FROM gold_vendor_risk_classification
     WHERE vendor_risk = 'Medium Risk') AS medium_risk_vendors,

    (SELECT COUNT(*)
     FROM gold_vendor_risk_classification
     WHERE vendor_risk = 'Low Risk') AS low_risk_vendors;
```
