```python
# Function to Profile Dataset


def profile_dataset(df, table_name):
    
    print("=" * 60)
    print(f"Dataset : {table_name}")
    print("=" * 60)
    
    print(f"Total Rows    : {df.count()}")
    print(f"Total Columns : {len(df.columns)}")
    
    print("\nSchema")
    df.printSchema()
    
    print("\nSummary Statistics")
    df.describe().display()
```

```python
# Profile all Bronze datasets

profile_dataset(df_purchase_orders, "Purchase Orders")

profile_dataset(df_invoices, "Invoices")

profile_dataset(df_vendor_contracts, "Vendor Contracts")
```

```python
# Null Value Count function 

def check_nulls(df, table_name):
    
    print("=" * 60)
    print(f"Null Count : {table_name}")
    print("=" * 60)
    
    null_df = df.select([
        count(when(col(c).isNull(), c)).alias(c)
        for c in df.columns
    ])
    
    display(null_df)
```

```python
check_nulls(df_purchase_orders, "Purchase Orders")

check_nulls(df_invoices, "Invoices")

check_nulls(df_vendor_contracts, "Vendor Contracts")
```

```python

# Missing Value Percentage function

def missing_percentage(df, table_name):
    
    total_rows = df.count()
    
    print("=" * 60)
    print(f"Missing Percentage : {table_name}")
    print("=" * 60)
    
    missing_df = df.select([
        (
            count(when(col(c).isNull(), c)) / total_rows * 100
        ).alias(c)
        for c in df.columns
    ])
    
    display(missing_df)
```

```python
missing_percentage(df_purchase_orders, "Purchase Orders")

missing_percentage(df_invoices, "Invoices")

missing_percentage(df_vendor_contracts, "Vendor Contracts")
```

```python
# ==========================================================
# Remove Records with Missing Business Keys
# ==========================================================

# Purchase Orders
df_purchase_orders = df_purchase_orders.filter(
    col("po_id").isNotNull() &
    col("vendor_id").isNotNull()
)

# Invoices
df_invoices = df_invoices.filter(
    col("invoice_id").isNotNull() &
    col("po_id").isNotNull() &
    col("vendor_id").isNotNull()
)

# Vendor Contracts
df_vendor_contracts = df_vendor_contracts.filter(
    col("contract_id").isNotNull() &
    col("vendor_id").isNotNull()
)

print("Records with missing business keys removed successfully.")
```

```python
# only unique and the primary key duplicates in the reocrd must be deal,
# Check Duplicate Business Keys


def check_duplicates(df, key_column, table_name):

    duplicate_count = (
        df.groupBy(key_column)
          .count()
          .filter(col("count") > 1)
          .count()
    )

    print(f"{table_name}")
    print(f"Duplicate {key_column}: {duplicate_count}")
```

```python
# Purchase Orders
check_duplicates(df_purchase_orders, "po_id", "Purchase Orders")

# Invoices
check_duplicates(df_invoices, "invoice_id", "Invoices")

# Vendor Contracts (Exact Duplicate Rows)
vendor_contract_duplicates = (
    df_vendor_contracts.count() - df_vendor_contracts.dropDuplicates().count()
)

print("Vendor Contracts")
print(f"Exact Duplicate Rows : {vendor_contract_duplicates}")
```

```python
# Remove Duplicate Records
# Purchase Orders
df_purchase_orders = df_purchase_orders.dropDuplicates(["po_id"])

# Invoices
df_invoices = df_invoices.dropDuplicates(["invoice_id"])

# Vendor Contracts
# Remove only exact duplicate rows to preserve historical records
# required for future SCD Type 2 implementation.
df_vendor_contracts = df_vendor_contracts.dropDuplicates()

print("Duplicate records removed successfully.")
```

```python
print("Duplicate Validation After Cleaning")
# Purchase Orders
check_duplicates(df_purchase_orders, "po_id", "Purchase Orders")
# Invoices
check_duplicates(df_invoices, "invoice_id", "Invoices")
# Vendor Contracts
vendor_contract_duplicates = (
    df_vendor_contracts.count() - df_vendor_contracts.dropDuplicates().count()
)

print("Vendor Contracts")
print(f"Exact Duplicate Rows : {vendor_contract_duplicates}")
```

```python
# ==========================================================
# Standardize Purchase Orders
# ==========================================================

before_count = df_purchase_orders.count()

df_purchase_orders = (
    df_purchase_orders
    .withColumn("vendor_id", upper(trim(col("vendor_id"))))
    .withColumn("item_name", initcap(trim(col("item_name"))))
    .withColumn("region", upper(trim(col("region"))))
    .withColumn("po_date", expr("try_cast(po_date AS DATE)"))
    .filter(col("po_date").isNotNull())
)

after_count = df_purchase_orders.count()

print(f"Purchase Orders - Invalid date records removed: {before_count - after_count}")
```

```python
# ==========================================================
# Standardize Invoices
# ==========================================================

before_count = df_invoices.count()

df_invoices = (
    df_invoices
    .withColumn("vendor_id", upper(trim(col("vendor_id"))))
    .withColumn("payment_status", upper(trim(col("payment_status"))))
    .withColumn("invoice_date", expr("try_cast(invoice_date AS DATE)"))
    .filter(col("invoice_date").isNotNull())
)

after_count = df_invoices.count()

print(f"Invoices - Invalid date records removed: {before_count - after_count}")
```

```python
# ==========================================================
# Standardize Vendor Contracts
# ==========================================================

before_count = df_vendor_contracts.count()

df_vendor_contracts = (
    df_vendor_contracts
    .withColumn("vendor_id", upper(trim(col("vendor_id"))))
    .withColumn("item_name", initcap(trim(col("item_name"))))
    .withColumn("category", initcap(trim(col("category"))))
    .withColumn("payment_terms", upper(trim(col("payment_terms"))))
    .withColumn("effective_date", expr("try_cast(effective_date AS DATE)"))
    .filter(col("effective_date").isNotNull())
)

after_count = df_vendor_contracts.count()

print(f"Vendor Contracts - Invalid date records removed: {before_count - after_count}")
```

```python

# Missing Values


def check_missing_values(df, table_name):

    print(f"\n{table_name}")

    display(
        df.select([
            count(when(col(c).isNull(), c)).alias(c)
            for c in df.columns
        ])
    )
```

```python
check_missing_values(df_purchase_orders,"Purchase Orders")

check_missing_values(df_invoices,"Invoices")

check_missing_values(df_vendor_contracts,"Vendor Contracts")
```

```python

# Numeric Column Statistics


def numeric_statistics(df,column):

    stats = df.select(
        mean(column).alias("Mean"),
        percentile_approx(column,0.5).alias("Median"),
        mode(column).alias("Mode"),
        skewness(column).alias("Skewness")
    )

    display(stats)
```

```python
numeric_statistics(df_purchase_orders,"quantity")

numeric_statistics(df_invoices,"invoiced_amount")

numeric_statistics(df_vendor_contracts,"contract_price")
```

```python
# ==========================================================
# Fill Missing Numeric Values
# ==========================================================

# Purchase Orders - Mean
quantity_mean = df_purchase_orders.select(mean("quantity")).first()[0]

df_purchase_orders = df_purchase_orders.fillna({
    "quantity": quantity_mean
})

# Invoices - Median
invoice_median = df_invoices.select(
    percentile_approx("invoiced_amount", 0.5)
).first()[0]

df_invoices = df_invoices.fillna({
    "invoiced_amount": invoice_median
})

# Vendor Contracts - Mean
contract_price_mean = df_vendor_contracts.select(
    mean("contract_price")
).first()[0]

df_vendor_contracts = df_vendor_contracts.fillna({
    "contract_price": contract_price_mean
})

print("Numeric missing values handled successfully.")
```

```python

# Fill Missing Values using Mode


def fill_mode(df, column_name):

    mode_row = (
        df.filter(col(column_name).isNotNull())
          .groupBy(column_name)
          .count()
          .orderBy(desc("count"))
          .first()
    )

    if mode_row:
        mode_value = mode_row[0]
        return df.fillna({column_name: mode_value})

    return df
```

```python
# Purchase Orders
df_purchase_orders = fill_mode(df_purchase_orders, "vendor_id")
df_purchase_orders = fill_mode(df_purchase_orders, "item_name")
df_purchase_orders = fill_mode(df_purchase_orders, "region")

# Invoices
df_invoices = fill_mode(df_invoices, "vendor_id")
df_invoices = fill_mode(df_invoices, "payment_status")

# Vendor Contracts
df_vendor_contracts = fill_mode(df_vendor_contracts, "vendor_id")
df_vendor_contracts = fill_mode(df_vendor_contracts, "item_name")
df_vendor_contracts = fill_mode(df_vendor_contracts, "category")
df_vendor_contracts = fill_mode(df_vendor_contracts, "payment_terms")

print("Categorical missing values handled successfully.")
```

```python
# ==========================================================
# Purchase Orders Validation
# ==========================================================

before_count = df_purchase_orders.count()

# Quantity must be greater than 0
df_purchase_orders = df_purchase_orders.filter(col("quantity") > 0)

after_count = df_purchase_orders.count()

print(f"Invalid Purchase Orders Removed : {before_count - after_count}")
```

```python
# ==========================================================
# Invoice Validation
# ==========================================================

before_count = df_invoices.count()

# Invoice amount must be greater than 0
df_invoices = df_invoices.filter(col("invoiced_amount") > 0)

after_count = df_invoices.count()

print(f"Invalid Invoices Removed : {before_count - after_count}")
```

```python
# ==========================================================
# Invoice Validation
# ==========================================================

before_count = df_invoices.count()

# Invoice amount must be greater than 0
df_invoices = df_invoices.filter(col("invoiced_amount") > 0)

after_count = df_invoices.count()

print(f"Invalid Invoices Removed : {before_count - after_count}")
```

```python
# ==========================================================
# Vendor Contract Validation
# ==========================================================

before_count = df_vendor_contracts.count()

# Contract price must be greater than 0
df_vendor_contracts = df_vendor_contracts.filter(col("contract_price") > 0)

after_count = df_vendor_contracts.count()

print(f"Invalid Contracts Removed : {before_count - after_count}")
```

```python
# ==========================================================
# Referential Integrity
# ==========================================================

# Invoice should have a matching Purchase Order

df_invoices = df_invoices.join(
    df_purchase_orders.select("po_id"),
    on="po_id",
    how="inner"
)

print("Invoice - Purchase Order validation completed.")
```

```python
# ==========================================================
# Vendor Validation
# ==========================================================

# Vendors in Purchase Orders must exist in Vendor Contracts

valid_vendors = df_vendor_contracts.select("vendor_id").distinct()

df_purchase_orders = df_purchase_orders.join(
    valid_vendors,
    on="vendor_id",
    how="inner"
)

df_invoices = df_invoices.join(
    valid_vendors,
    on="vendor_id",
    how="inner"
)

print("Vendor validation completed.")
```

```python
print("Business Rule Validation Completed Successfully.")
```

```python
# ==========================================================
# Add Silver Layer Metadata
# ==========================================================

df_purchase_orders = (
    df_purchase_orders
    .withColumn("silver_load_timestamp", current_timestamp())
)

df_invoices = (
    df_invoices
    .withColumn("silver_load_timestamp", current_timestamp())
)

df_vendor_contracts = (
    df_vendor_contracts
    .withColumn("silver_load_timestamp", current_timestamp())
)

print("Silver metadata added successfully.")
```

```python
# ==========================================================
# Write Silver Delta Tables
# ==========================================================

df_purchase_orders.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("silver_purchase_orders")

df_invoices.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("silver_invoices")

df_vendor_contracts.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("silver_vendor_contracts")

print("Silver Delta tables created successfully.")
```
