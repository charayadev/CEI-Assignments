## What are we doing?

Before creating Gold tables, we will use the cleaned Silver Delta tables as the source for all analytical datasets.

## Why are we doing it?

The Silver Layer already contains data that has been cleaned, standardized, validated, and transformed according to business rules. Using these tables ensures that the Gold Layer is built on trusted and consistent data.

If we directly used raw CSV files, we would need to repeat all the data cleaning steps again, increasing complexity and the possibility of inconsistent results.


```python
# Create Temporary Views for Spark SQL

purchase_orders_df.createOrReplaceTempView("purchase_orders")

invoices_df.createOrReplaceTempView("invoices")

vendor_contracts_df.createOrReplaceTempView("vendor_contracts")

print("Temporary views created successfully.")
```

```python
historical_procurement_df = spark.sql("""

SELECT

    po.po_id,
    po.vendor_id,
    po.item_name,
    po.quantity,
    po.region,
    po.po_date,

    inv.invoice_id,
    inv.invoiced_amount,
    inv.invoice_date,
    inv.payment_status,

    vc.contract_id,
    vc.category,
    vc.contract_price,
    vc.payment_terms,
    vc.start_date,
    vc.end_date,
    vc.version_number

FROM purchase_orders po

LEFT JOIN invoices inv
ON po.po_id = inv.po_id

LEFT JOIN vendor_contracts vc
ON po.vendor_id = vc.vendor_id
AND po.item_name = vc.item_name
AND po.po_date BETWEEN vc.start_date
                   AND COALESCE(vc.end_date, DATE('9999-12-31'))

""")

print("Historical procurement dataset created successfully.")
```

```python
display(historical_procurement_df.limit(5))
```

```python

# Gold Table 1 – Vendor Spend Summary

## What are we doing?

# In this section, we will create the first Gold table named **gold_vendor_spend_summary**.

# This table provides a summarized view of procurement spending for each vendor and item. Instead of looking at individual purchase orders, business users can quickly understand overall vendor performance.

## Why are we doing it?

# Procurement teams frequently need answers to questions such as:

# - How much have we spent with each vendor?
# - How many purchase orders has each vendor received?
# - What is the average invoice amount?
# - What is the latest contract price for a product?

# Rather than calculating these metrics every time, the Gold Layer stores them in a business-ready format.


gold_vendor_spend_summary = (
    historical_procurement_df
    .fillna({
        "invoiced_amount": 0,
        "contract_price": 0
    })
    .groupBy("vendor_id", "item_name")
    .agg(
        F.sum("invoiced_amount").alias("total_spend"),
        F.avg("invoiced_amount").alias("average_invoice_amount"),
        F.countDistinct("po_id").alias("total_purchase_orders"),
        F.sum("quantity").alias("total_quantity_purchased"),
        F.max("contract_price").alias("latest_active_contract_price")
    )
)
```

```python
# Calculate Total Number of Vendors

total_vendors = purchase_orders_df.select("vendor_id").distinct().count()

# Add Total Vendors as a Column

gold_vendor_spend_summary = gold_vendor_spend_summary.withColumn(
    "total_vendors",
    F.lit(total_vendors)
)

print("Total Vendors column added successfully.")
```

```python
# Add Gold Layer Load Timestamp

gold_vendor_spend_summary = gold_vendor_spend_summary.withColumn(
    "gold_load_timestamp",
    F.current_timestamp()
)

print("Gold load timestamp added successfully.")
```

```python
print("Gold Vendor Spend Summary Schema")

gold_vendor_spend_summary.printSchema()

print(f"Total Records: {gold_vendor_spend_summary.count()}")

display(gold_vendor_spend_summary.limit(10))
```

```python
null_summary = gold_vendor_spend_summary.select([
    F.count(F.when(F.col(c).isNull(), c)).alias(c)
    for c in gold_vendor_spend_summary.columns
])

display(null_summary)
```

```python
gold_vendor_spend_summary.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("gold_vendor_spend_summary")

print("Gold Vendor Spend Summary saved successfully.")
```

# Gold Table 2 – Price Variance Report

## What are we doing?

In this section, we will create the **gold_price_variance_report** table.

This table compares the amount that should have been paid according to the vendor contract with the actual invoice amount.


```python
historical_procurement_clean = historical_procurement_df.fillna({
    "contract_price": 0,
    "invoiced_amount": 0
})

historical_procurement_clean.createOrReplaceTempView("historical_procurement_clean")

print("Historical procurement dataset cleaned successfully.")
```

```python
gold_price_variance_report = spark.sql("""

SELECT

    vendor_id,
    po_id,
    item_name,
    quantity,
 


    contract_price,

    invoiced_amount AS actual_amount,

    (contract_price * quantity) AS expected_amount,

    (invoiced_amount - (contract_price * quantity)) AS variance,

    CASE
    WHEN (contract_price * quantity) = 0 THEN 0
    ELSE ROUND(
        (
            (invoiced_amount - (contract_price * quantity))
            / (contract_price * quantity)
        ) * 100,
        2
    )
END AS variance_percentage,
    CASE

        WHEN invoiced_amount > (contract_price * quantity)
            THEN 'Overcharged'

        WHEN invoiced_amount < (contract_price * quantity)
            THEN 'Undercharged'

        ELSE 'Within Contract'

    END AS price_status

FROM historical_procurement_clean

""")

print("Gold Price Variance Report created successfully.")
```

```python
gold_price_variance_report = gold_price_variance_report.withColumn(
    "gold_load_timestamp",
    F.current_timestamp()
)

print("Gold load timestamp added successfully.")
```

```python
# Save Gold Price Variance Report

gold_price_variance_report.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("gold_price_variance_report")

print("Gold Price Variance Report saved successfully.")
```

# Gold Table 3 – Vendor Risk Classification

## What are we doing?

In this section, we will classify vendors into different risk categories based on their pricing behavior and invoice payment status.

## Why are we doing it?

Not every vendor poses the same level of business risk. Procurement teams monitor vendor performance to identify suppliers who frequently overcharge or have payment-related issues.


```python
# Create Temporary Views

gold_price_variance_report.createOrReplaceTempView("price_variance_report")

invoices_df.createOrReplaceTempView("silver_invoices")

print("Temporary views created successfully.")
```

```python
gold_price_variance_report_df = spark.table("gold_price_variance_report")
gold_price_variance_report.createOrReplaceTempView("price_variance_report")
```

```python
gold_vendor_risk_classification = spark.sql("""

WITH vendor_metrics AS (

SELECT

    p.vendor_id,

    ROUND(AVG(p.variance_percentage),2) AS average_variance_percentage,

    MAX(
        CASE
            WHEN i.payment_status = 'OVERDUE' THEN 1
            ELSE 0
        END
    ) AS has_overdue_invoice

FROM price_variance_report p

LEFT JOIN silver_invoices i
ON p.po_id = i.po_id

GROUP BY p.vendor_id

)

SELECT

    vendor_id,

    average_variance_percentage,

    has_overdue_invoice,

    CASE

        WHEN average_variance_percentage > 10
             AND has_overdue_invoice = 1
        THEN 'High Risk'

        WHEN average_variance_percentage > 5
             OR has_overdue_invoice = 1
        THEN 'Medium Risk'

        ELSE 'Low Risk'

    END AS vendor_risk

FROM vendor_metrics

""")

print("Vendor Risk Classification created successfully.")
```

```python
gold_vendor_risk_classification = gold_vendor_risk_classification.withColumn(
    "gold_load_timestamp",
    F.current_timestamp()
)

print("Gold load timestamp added successfully.")
```

```python
# Save Gold Vendor Risk Classification

gold_vendor_risk_classification.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("gold_vendor_risk_classification")

print("Gold Vendor Risk Classification saved successfully.")
```

# Gold Table 4 – Regional Spend Analysis

## What are we doing?

In this section, we will create the final Gold table named **gold_regional_spend_analysis**.

## Why are we doing it?

Procurement managers often need to compare procurement spending across different regions and identify the highest-spending vendors within each region.

```python
regional_vendor_spend = historical_procurement_clean.groupBy(
    "region",
    "vendor_id"
).agg(
    F.sum("invoiced_amount").alias("total_spend"),
    F.avg("invoiced_amount").alias("average_spend"),
    F.countDistinct("po_id").alias("total_purchase_orders")
)
```

```python
window_spec = Window.partitionBy("region").orderBy(F.desc("total_spend"))

gold_regional_spend_analysis = regional_vendor_spend.withColumn(
    "vendor_rank",
    F.dense_rank().over(window_spec)
)
```

```python
regional_vendor_count = historical_procurement_clean.groupBy("region").agg(
    F.countDistinct("vendor_id").alias("total_vendors")
)

gold_regional_spend_analysis = (
    gold_regional_spend_analysis
    .join(regional_vendor_count, "region", "left")
    .withColumn("gold_load_timestamp", F.current_timestamp())
)

print("Regional Spend Analysis created successfully.")
```

```python
# Save Gold Regional Spend Analysis

gold_regional_spend_analysis.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("gold_regional_spend_analysis")

print("Gold Regional Spend Analysis saved successfully.")
```
