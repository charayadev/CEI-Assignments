# Procurement & Vendor Spend Analytics Pipeline
## Final Data Validation Report

### Project Overview
This notebook performs the final validation of the Procurement & Vendor Spend Analytics Pipeline developed using **Databricks**, **PySpark**, **Delta Lake**, and **Spark SQL**.

The purpose of this notebook is to verify the integrity and quality of the data stored in the Bronze, Silver, and Gold layers after the completion of the ETL pipeline.


```python
from pyspark.sql import functions as F
from pyspark.sql.types import *
from functools import reduce

# Tables to validate
tables = {
    "Bronze": [
        "bronze_purchase_orders",
        "bronze_invoices",
        "bronze_vendor_contracts"
    ],
    "Silver": [
        "silver_purchase_orders",
        "silver_invoices",
        "silver_vendor_contracts_history"
    ],
    "Gold": [
        "gold_vendor_spend_summary",
        "gold_price_variance_report",
        "gold_vendor_risk_classification",
        "gold_regional_spend_analysis"
    ]
}

validation_results = []
print("done")
```

```python
print("========== BRONZE LAYER VALIDATION ==========\n")

for table in tables["Bronze"]:

    print(f"\nValidating Table: {table}")

    df = spark.table(table)

    rows = df.count()
    cols = len(df.columns)

    print(f"Record Count : {rows}")
    print(f"Column Count : {cols}")

    print("\nSchema")
    df.printSchema()

    print("\nData Types")
    print(df.dtypes)

    null_df = df.select([
        F.count(F.when(F.col(c).isNull(), c)).alias(c)
        for c in df.columns
    ])

    total_nulls = sum(null_df.first())

    print("\nNull Values")
    null_df.show()

    duplicates = rows - df.dropDuplicates().count()
    print(f"Duplicate Rows : {duplicates}")

    print("\nSample Records")
    df.show(5, truncate=False)

    status = "PASS" if rows > 0 else "FAIL"

    validation_results.append((
        table, "Bronze", rows, cols,
        total_nulls, duplicates,
        status, "Validated Successfully"
    ))

    print(f"Validation Status : {status}")
```
