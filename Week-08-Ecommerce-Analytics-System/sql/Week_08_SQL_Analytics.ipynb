```sql
-- revenue by per customer

SELECT
    c.customer_id,
    c.customer_name,
    c.customer_type,
    COUNT(DISTINCT o.order_id) AS total_orders,
    ROUND(SUM(oi.total_amount),2) AS total_revenue
FROM customers_clean c
JOIN orders_clean o
    ON c.customer_id = o.customer_id
JOIN order_items_clean oi
    ON o.order_id = oi.order_id
GROUP BY
    c.customer_id,
    c.customer_name,
    c.customer_type
ORDER BY total_revenue DESC;
```

```sql
-- revenue by product category

SELECT
    p.category,
    ROUND(SUM(oi.total_amount),2) AS total_revenue
FROM products_clean p
JOIN order_items_clean oi
    ON p.product_id = oi.product_id
GROUP BY p.category
ORDER BY total_revenue DESC;
```

```sql

--Monthly Revenue Trend
SELECT
    o.order_month_year,
    ROUND(SUM(oi.total_amount),2) AS monthly_revenue
FROM orders_clean o
JOIN order_items_clean oi
ON o.order_id = oi.order_id
GROUP BY o.order_month_year
ORDER BY o.order_month_year;
```

```sql
-- Top 10 Products by Revenue
SELECT
    p.product_name,
    p.category,
    ROUND(SUM(oi.total_amount),2) AS revenue
FROM products_clean p
JOIN order_items_clean oi
ON p.product_id = oi.product_id
GROUP BY
    p.product_name,
    p.category
ORDER BY revenue DESC
LIMIT 10;
```

```sql
-- Top 10 Products by Quantity Sold
SELECT
    p.product_name,
    p.category,
    SUM(oi.quantity) AS total_quantity
FROM products_clean p
JOIN order_items_clean oi
ON p.product_id = oi.product_id
GROUP BY
    p.product_name,
    p.category
ORDER BY total_quantity DESC
LIMIT 10;
```

```sql
-- Average Order Value (AOV)
SELECT
    c.customer_type,
    ROUND(AVG(order_total),2) AS average_order_value
FROM
(
    SELECT
        o.order_id,
        o.customer_id,
        SUM(oi.total_amount) AS order_total
    FROM orders_clean o
    JOIN order_items_clean oi
    ON o.order_id = oi.order_id
    GROUP BY
        o.order_id,
        o.customer_id
) temp
JOIN customers_clean c
ON temp.customer_id = c.customer_id
GROUP BY c.customer_type;
```

```sql
-- Rank Customers by Lifetime Value
SELECT
    c.customer_id,
    c.customer_name,
    ROUND(SUM(oi.total_amount),2) AS lifetime_value,

    RANK() OVER(
        ORDER BY SUM(oi.total_amount) DESC
    ) AS customer_rank

FROM customers_clean c

JOIN orders_clean o
ON c.customer_id = o.customer_id

JOIN order_items_clean oi
ON o.order_id = oi.order_id

GROUP BY
c.customer_id,
c.customer_name

ORDER BY customer_rank;
```

```sql
-- Dense Rank Customers
SELECT
    c.customer_id,
    c.customer_name,
    ROUND(SUM(oi.total_amount),2) AS lifetime_value,

    DENSE_RANK() OVER(
        ORDER BY SUM(oi.total_amount) DESC
    ) AS dense_rank

FROM customers_clean c

JOIN orders_clean o
ON c.customer_id = o.customer_id

JOIN order_items_clean oi
ON o.order_id = oi.order_id

GROUP BY
c.customer_id,
c.customer_name

ORDER BY dense_rank;
```

```sql
-- Running Revenue
SELECT
    o.order_month_year,

    ROUND(
        SUM(oi.total_amount),
        2
    ) AS monthly_revenue,

    ROUND(
        SUM(
            SUM(oi.total_amount)
        ) OVER(
            ORDER BY o.order_month_year
        ),
        2
    ) AS running_total

FROM orders_clean o

JOIN order_items_clean oi
ON o.order_id = oi.order_id

GROUP BY
o.order_month_year

ORDER BY
o.order_month_year;
```

```sql
-- Moving Average
SELECT
    order_month_year,
    monthly_revenue,

    ROUND(
        AVG(monthly_revenue)
        OVER(
            ORDER BY order_month_year
            ROWS BETWEEN 2 PRECEDING
            AND CURRENT ROW
        ),
        2
    ) AS moving_average

FROM
(
SELECT
    o.order_month_year,
    SUM(oi.total_amount) AS monthly_revenue

FROM orders_clean o

JOIN order_items_clean oi
ON o.order_id = oi.order_id

GROUP BY
o.order_month_year
) revenue;
```

```sql
-- CTE (Monthly Revenue)
WITH MonthlyRevenue AS
(
SELECT

o.order_month_year,

ROUND(
SUM(oi.total_amount),
2
) AS revenue

FROM orders_clean o

JOIN order_items_clean oi

ON o.order_id=oi.order_id

GROUP BY
o.order_month_year
)

SELECT *

FROM MonthlyRevenue

ORDER BY order_month_year;
```

```sql
-- Monthly Growth Rate (CTE)
WITH MonthlyRevenue AS
(
SELECT

o.order_month_year,

ROUND(
SUM(oi.total_amount),
2
) AS revenue

FROM orders_clean o

JOIN order_items_clean oi

ON o.order_id=oi.order_id

GROUP BY
o.order_month_year
)

SELECT

order_month_year,

revenue,

LAG(revenue)
OVER(
ORDER BY order_month_year
)
AS previous_month,

ROUND(
(
revenue-
LAG(revenue)
OVER(
ORDER BY order_month_year
)
)
/
LAG(revenue)
OVER(
ORDER BY order_month_year
)
*100,
2
)
AS growth_percentage

FROM MonthlyRevenue;
```

```sql
-- Customer First Purchase (Cohort)
WITH FirstPurchase AS
(
SELECT
customer_id,
MIN(order_date) AS first_purchase
FROM orders_clean
GROUP BY customer_id
)

SELECT
customer_id,
DATE_FORMAT(first_purchase,'yyyy-MM') AS cohort_month
FROM FirstPurchase
ORDER BY cohort_month;
```

```sql
-- Cohort Size
WITH FirstPurchase AS
(
SELECT
customer_id,
MIN(order_date) AS first_purchase
FROM orders_clean
GROUP BY customer_id
)

SELECT
DATE_FORMAT(first_purchase,'yyyy-MM') AS cohort_month,
COUNT(customer_id) AS total_customers
FROM FirstPurchase
GROUP BY DATE_FORMAT(first_purchase,'yyyy-MM')
ORDER BY cohort_month;
```

```sql
-- Repeat Customers
SELECT

customer_id,

COUNT(order_id) AS total_orders,

CASE

WHEN COUNT(order_id)=1
THEN 'One-Time'

ELSE 'Repeat'

END AS customer_status

FROM orders_clean

GROUP BY customer_id

ORDER BY total_orders DESC;
```

```sql
--Customer Retention Summary

SELECT

COUNT(DISTINCT customer_id) AS total_customers,

COUNT(
DISTINCT
CASE
WHEN total_orders>1
THEN customer_id
END
) AS retained_customers,

ROUND(

COUNT(
DISTINCT
CASE
WHEN total_orders>1
THEN customer_id
END
)

*100.0/

COUNT(DISTINCT customer_id),

2

) AS retention_rate

FROM

(
SELECT
customer_id,
COUNT(order_id) AS total_orders
FROM orders_clean
GROUP BY customer_id
)t;
```

```sql
-- Churned Customers
SELECT

customer_id,

COUNT(order_id) AS total_orders,

CASE

WHEN COUNT(order_id)=1
THEN 'Churned'

ELSE 'Active'

END AS customer_status

FROM orders_clean

GROUP BY customer_id

ORDER BY total_orders;
```

```sql
-- Purchase Frequency Segmentation
SELECT

customer_id,

COUNT(order_id) AS total_orders,

CASE

WHEN COUNT(order_id)=1 THEN 'One-Time'

WHEN COUNT(order_id) BETWEEN 2 AND 5 THEN 'Occasional'

ELSE 'Loyal'

END AS frequency_segment

FROM orders_clean

GROUP BY customer_id

ORDER BY total_orders DESC;
```

```sql
-- Spend Tier Segmentation
SELECT

c.customer_id,

c.customer_name,

ROUND(SUM(oi.total_amount),2) AS total_spend,

CASE

WHEN SUM(oi.total_amount)<50000
THEN 'Low'

WHEN SUM(oi.total_amount)<150000
THEN 'Medium'

ELSE 'High'

END AS spend_segment

FROM customers_clean c

JOIN orders_clean o
ON c.customer_id=o.customer_id

JOIN order_items_clean oi
ON o.order_id=oi.order_id

GROUP BY

c.customer_id,
c.customer_name

ORDER BY total_spend DESC;
```

```sql
-- RFM Metrics
SELECT

o.customer_id,

MAX(o.order_date) AS last_purchase,

DATEDIFF(
CURRENT_DATE(),
MAX(o.order_date)
) AS recency,

COUNT(DISTINCT o.order_id) AS frequency,

ROUND(SUM(oi.total_amount),2) AS monetary

FROM orders_clean o

JOIN order_items_clean oi

ON o.order_id=oi.order_id

GROUP BY o.customer_id;
```

```sql
--Complete RFM Segmentation
WITH RFM AS
(
SELECT

o.customer_id,

DATEDIFF(
CURRENT_DATE(),
MAX(o.order_date)
) AS recency,

COUNT(DISTINCT o.order_id) AS frequency,

ROUND(SUM(oi.total_amount),2) AS monetary

FROM orders_clean o

JOIN order_items_clean oi

ON o.order_id=oi.order_id

GROUP BY o.customer_id
)

SELECT

*,

CASE

WHEN frequency>=10
AND monetary>=150000
THEN 'VIP'

WHEN frequency>=5
THEN 'Loyal'

WHEN frequency>=2
THEN 'Regular'

ELSE 'New'

END AS customer_segment

FROM RFM

ORDER BY monetary DESC;
```
