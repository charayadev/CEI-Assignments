```python
# the customers table has null values in phone column, so we will replace them with "Not Available and save it as customers_clean"
customers_clean = customers.fillna({
    "phone": "Not Available"
})
```

```python
# prouduct table is perfect with the null values no, null values are there
products_clean = products
```

```python

# orders table has null values in cutsomer id column, so we will remove the rows with null values as we can not fill them random values 
orders_clean = orders.filter(col("customer_id").isNotNull())

```

```python
# order items have no null values save it as it
order_items_clean = order_items
```

```python
from pyspark.sql.functions import count

def check_duplicates(df, table_name):
    print(f"\n{'='*50}")
    print(f"Checking Duplicate Values in {table_name}")
    print(f"{'='*50}")

    for col_name in df.columns:
        duplicate_count = (
            df.groupBy(col_name)
              .count()
              .filter("count > 1")
              .count()
        )

        print(f"{col_name}: {duplicate_count} duplicate values")
```

```python
check_duplicates(customers_clean, "Customers")
check_duplicates(products_clean, "Products")
check_duplicates(orders_clean, "Orders")
check_duplicates(order_items_clean, "Order Items")
```

```python
# since all the major columns are already unique we will move forwards
```

```python
# check the foregin key and primary key in another table realtion orders and customers  
invalid_customers = (
    orders_clean.join(
        customers_clean,
        "customer_id",
        "left_anti"
    )
)

print("Invalid Customer IDs:", invalid_customers.count())

display(invalid_customers)
```

```python
# check the foregin key and primary key in another table realtion produtcs and order items
invalid_products = (
    order_items_clean.join(
        products_clean,
        "product_id",
        "left_anti"
    )
)

print("Invalid Product IDs:", invalid_products.count())

display(invalid_products)
```

```python
#  found some of the order _items invalid id and fix it on product id
order_items_clean = order_items_clean.join(
    products_clean.select("product_id"),
    on="product_id",
    how="inner"
)
```

```python
invalid_orders = (
    order_items_clean.join(
        orders_clean,
        "order_id",
        "left_anti"
    )
)

print("Invalid Order IDs:", invalid_orders.count())

display(invalid_orders)
```

```python
#  found some of the order _items invalid id and fix it on order_id
order_items_clean = order_items_clean.join(
    orders_clean.select("order_id"),
    on="order_id",
    how="inner"
)
```

```python
order_items_clean = order_items_clean.withColumn(
    "total_amount",
    round(
        col("quantity") *
        col("unit_price") *
        (1 - col("discount_percent")/100),
        2
    )
)
```

```python
orders_clean = orders.withColumn(
    "order_date",
    when(
        col("order_date").contains("/"),
        to_date(col("order_date"), "dd/MM/yyyy")
    ).otherwise(
        to_date(col("order_date"), "yyyy-MM-dd")
    )
)

orders_clean = (
    orders_clean
    .withColumn("order_year", year("order_date"))
    .withColumn("order_month", month("order_date"))
    .withColumn("order_month_year", date_format("order_date", "yyyy-MM"))
)
```

```python
customers_clean.write.mode("overwrite").saveAsTable("customers_clean")

products_clean.write.mode("overwrite").saveAsTable("products_clean")

orders_clean.write.mode("overwrite").saveAsTable("orders_clean")

order_items_clean.write.mode("overwrite").saveAsTable("order_items_clean")
```
